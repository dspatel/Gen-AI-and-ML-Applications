```python
#connect to tradingview and get yesterdays 5 minute candlestick data for NVDA
import tvDatafeed
from tvDatafeed import TvDatafeed, Interval
import keyring
import logging
from logging import getLogger


user = keyring.get_password("tradingview", "username")
pwd = keyring.get_password("tradingview", "password")
print("Logging in to TradingView as", user)
print("Password :", pwd)
tv=TvDatafeed(username=user, password=pwd)
logger = getLogger(__name__)
logger.info("✅ tvDatafeed logged in using OS keychain.")
data = tv.get_hist(symbol="NVDA", exchange="NASDAQ", interval=Interval.in_5_minute, n_bars=100)
#print all data without truncation
import pandas as pd
pd.set_option('display.max_rows', None)
print(data)
```
